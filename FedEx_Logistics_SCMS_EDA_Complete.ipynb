## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Parse dates and numeric measures into analysis columns while preserving every raw source field.
df = fedex_df.copy()
numeric_columns = {
    "Line Item Quantity": "line_item_quantity",
    "Line Item Value": "line_item_value_usd",
    "Pack Price": "pack_price_usd",
    "Unit Price": "unit_price_usd",
    "Unit of Measure (Per Pack)": "unit_measure_per_pack",
    "Weight (Kilograms)": "weight_kg",
    "Freight Cost (USD)": "freight_cost_usd",
    "Line Item Insurance (USD)": "line_item_insurance_usd",
}
date_columns = {
    "PQ First Sent to Client Date": "pq_first_sent_to_client_date",
    "PO Sent to Vendor Date": "po_sent_to_vendor_date",
    "Scheduled Delivery Date": "scheduled_delivery_date",
    "Delivered to Client Date": "delivered_to_client_date",
    "Delivery Recorded Date": "delivery_recorded_date",
}

# Coerce numeric values; narrative markers remain in fedex_df and become NaN only in derived measures.
for source_col, analysis_col in numeric_columns.items():
    df[analysis_col] = pd.to_numeric(df[source_col], errors="coerce")

# Parse source dates. Text such as process labels and "Date Not Captured" becomes NaT for calculations.
for source_col, analysis_col in date_columns.items():
    df[analysis_col] = pd.to_datetime(df[source_col], errors="coerce", dayfirst=True)

# Positive delay means late; zero means exactly on schedule; negative means early.
df["delivery_delay_days"] = (
    df["delivered_to_client_date"] - df["scheduled_delivery_date"]
).dt.days
df["delivery_status"] = np.select(
    [df["delivery_delay_days"].gt(0), df["delivery_delay_days"].eq(0), df["delivery_delay_days"].lt(0)],
    ["Late", "On schedule", "Early"],
    default="Unknown",
)
df["on_time_or_early"] = df["delivery_delay_days"].le(0).astype("boolean")
df.loc[df["delivery_delay_days"].isna(), "on_time_or_early"] = pd.NA

# This measures the PO-to-planned-delivery window, not physical transit time.
df["po_to_scheduled_days"] = (
    df["scheduled_delivery_date"] - df["po_sent_to_vendor_date"]
).dt.days
lead_bins = [-np.inf, -0.001, 30, 60, 90, 180, 365, np.inf]
lead_labels = ["Negative (review)", "0-30 days", "31-60 days", "61-90 days", "91-180 days", "181-365 days", ">365 days"]
df["po_schedule_lead_bucket"] = pd.cut(
    df["po_to_scheduled_days"], bins=lead_bins, labels=lead_labels, right=True
)

# Normalize freight by line-item value only where value is positive and freight is numeric.
df["freight_share_of_line_value_pct"] = (
    df["freight_cost_usd"] / df["line_item_value_usd"] * 100
).where(df["line_item_value_usd"].gt(0))

# Make availability explicit; no rows are dropped or imputed.
availability = pd.DataFrame({
    "numeric_values_available": {name: int(df[col].notna().sum()) for name, col in numeric_columns.items()},
    "date_values_parseable": {name: int(df[col].notna().sum()) for name, col in date_columns.items()},
})
display(availability)
print(f"Rows retained: {len(df):,}; no rows were dropped or imputed.")
print(f"Records with both scheduled and delivered dates: {df['delivery_delay_days'].notna().sum():,}")
````

### What all manipulations have you done and insights you found?

### What all manipulations have you done and insights you found?

- Preserved fedex_df and all original text values. Created a separate working copy, df.
- Parsed the five date columns with invalid process/status text coerced to NaT; created numeric analysis columns with pd.to_numeric(errors='coerce'). The raw notes remain available for investigation.
- Derived delivery_delay_days = delivered date - scheduled date, with positive values late, zero on schedule, and negative values early. The delivery dates are parseable for all 10,324 records, so the baseline delivery denominator is 10,324.
- Derived po_to_scheduled_days to measure the planned PO-to-delivery window, not transport time. Negative windows are retained and marked for review; timing comparisons use non-negative windows where stated.
- Derived freight as a percentage of line value only when freight is numeric and line value is positive.
- Kept rows with missing/unparseable values. No mean/mode/zero imputation was applied because the source contains meaningful statuses such as “Weight Captured Separately,” “Freight Included in Commodity Cost,” and “Invoiced Separately.”
- No duplicate rows or repeated IDs were found. Rates are record-weighted, not shipment-level or order-level, and category comparisons are descriptive rather than causal.

#### Chart - 7 - On-time-or-early rate by shipment mode

````python
# Bivariate: compare delivery outcomes by mode and retain denominators in hover.
mode_perf = (df.dropna(subset=["Shipment Mode", "delivery_delay_days"])
    .assign(on_time=lambda frame: frame["delivery_delay_days"].le(0))
    .groupby("Shipment Mode", observed=True)
    .agg(records=("ID", "size"), on_time_rate_pct=("on_time", "mean")).reset_index())
mode_perf["on_time_rate_pct"] *= 100
fig = px.bar(mode_perf, x="Shipment Mode", y="on_time_rate_pct", custom_data=["records"],
             title="Delivered on or before schedule by shipment mode",
             labels={"on_time_rate_pct": "On-time-or-early records (%)"},
             color="on_time_rate_pct", color_continuous_scale="Tealgrn")
fig.update_traces(hovertemplate="%{x}<br>On-time or early: %{y:.1f}%<br>Records: %{customdata[0]:,}<extra></extra>")
fig.update_layout(yaxis_range=[0, 100], coloraxis_showscale=False)
fig.show()
print(f"Rows with missing Shipment Mode excluded: {df['Shipment Mode'].isna().sum():,}")
````

##### 1. Why did you pick the specific chart?

A rate chart directly answers whether outcomes vary by mode; hover counts prevent rates from being read without denominators.

##### 2. What is/are the insight(s) found from the chart?

Air has a 90.4% on-time-or-early rate (n=6,113), Air Charter 88.5% (n=650), Truck 83.9% (n=2,830), and Ocean 82.5% (n=371). The 360 unassigned modes are excluded.
